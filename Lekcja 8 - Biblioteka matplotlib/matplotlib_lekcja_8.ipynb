````python
import matplotlib.pyplot as plt
import numpy as np

# x_points = np.array([-3, -2, -1, 0, 1, 2, 3])
x_points = np.arange(-3, 4, 1) # start, stop, step

# y = 2*x -1
# y_points = [2*x-1 for x in x_points]
y_points = 2 * x_points - 1

plt.plot(x_points, y_points, c="r", marker="s", linestyle="--", label="y = 2x - 1")
plt.title("Wykres funkcji liniowej")
plt.legend()
plt.xlabel("Oś X")
plt.ylabel("Oś Y")
plt.grid()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(25) # 0 - 1
y = np.random.rand(25)

plt.scatter(x, y, color="green", marker='o', label="Losowe punkty")
plt.title("Losowe punkty")
plt.legend()
plt.xlabel("Oś X")
plt.ylabel("Oś Y")
plt.grid()
plt.show()
````

````python
import matplotlib.pyplot as plt

dobra_gra = ["grafika", "lobby", "odporna na bugi", "dynamika", "multiplayer"]

procenty = [30, 10, 20, 20, 20]

plt.pie(procenty, labels=dobra_gra, autopct='%1.1f%%', colors=["red", "green", "blue", "orange", "purple"])
plt.title("Co musi miec dobra gra?")
plt.show()
````

````python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap="viridis")
plt.show()

# plt.savefig("wykres3D.png")
````
